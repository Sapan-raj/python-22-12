## Files, exceptional handling, logging andmemory management Questions

1. Difference between interpreted and compiled languages
- Interpreted: Code is executed line-by-line (e.g., Python). Slower but easier for debugging.

- Compiled: Code is first converted into machine code (e.g., C++). Faster at runtime.



2. What is exception handling in Python?
- It is a method to handle runtime errors using try, except, finally, and else blocks to prevent crashes and handle problems gracefully.

3. Purpose of the finally block
- It executes regardless of whether an exception occurred. Commonly used to close files or release resources.

4. What is logging in Python?
- Logging records messages during code execution. It's useful for debugging and tracking issues using the logging module.

5. Significance of __del__ method
- This destructor method is called when an object is deleted. Used to free resources like closing files or network connections.

6. Difference between import and from ... import
- import module: Access functions via module.function().

- from module import function: Use function() directly.

7. Handling multiple exceptions
- Use multiple except blocks or a tuple:

```python
try:
    # code
except (TypeError, ValueError) as e:
    print(e)
```

8. Purpose of with statement
- Manages resources automatically. Ensures files are closed:

9. Difference between multithreading and multiprocessing
- Multithreading: Multiple threads in one process (shared memory).

- Multiprocessing: Multiple processes (separate memory) for true parallelism.

10. Advantages of logging
- Tracks issues without stopping code

- Helpful in debugging and audit

- Records time, level, and messages

11. Memory management in Python
- Python uses reference counting and garbage collection (GC) to manage memory automatically.



12. Steps in exception handling
- 1.  Use try block to write risky code.

- 2. Handle errors using except.

- 3. (Optional) Add else for code if no error.

- 4. (Optional) Use finally for cleanup.

13. Importance of memory management
- Efficient memory use avoids leaks and improves performance, especially in large applications.

14. Role of try and except
- Used to catch and handle exceptions gracefully so the program doesn’t crash.

15. Python’s garbage collection
- It uses reference counting and a cyclic garbage collector to free unused objects automatically.

16. Purpose of else in exception handling
- Runs only if no exception occurs in try block:

```python
try:
    x = 10
except:
    pass
else:
    print("No error occurred")
```

18. Difference between os.fork() and multiprocessing
- os.fork() (Unix-only): Clones the current process.

- multiprocessing: Cross-platform module for process-based parallelism.

19. Importance of closing a file
- Releases system resources and ensures data is properly saved.



20. Difference between file.read() and file.readline()

## Practical Questions – Python Code Examples

1. Open file and write string

```python
with open('example.txt', 'w') as f:
    f.write("Hello, world!")
```

2. Read and print file line by line

```python
with open('example.txt', 'r') as f:
    for line in f:
        print(line.strip())
```

3. Handle file not found

```python
try:
    with open('nofile.txt', 'r') as f:
        content = f.read()
except FileNotFoundError:
    print("File not found.")
```

4. Read from one file, write to another

```python
with open('source.txt', 'r') as src, open('destination.txt', 'w') as dst:
    dst.write(src.read())
```

5. Handle division by zero

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero.")
```

6. Log division by zero error

```python
import logging
logging.basicConfig(filename='app.log', level=logging.ERROR)
try:
    result = 10 / 0
except ZeroDivisionError as e:
    logging.error("Division by zero: %s", e)
```

7. Logging at different levels

```python
import logging
logging.basicConfig(level=logging.DEBUG)
logging.info("Info message")
logging.warning("Warning message")
logging.error("Error occurred")
```

8. Handle file opening error

```python
try:
    with open('file.txt', 'r') as f:
        print(f.read())
except IOError:
    print("File cannot be opened.")
```

9. Read file to list

```python
with open('file.txt', 'r') as f:
    lines = f.readlines()
```

10. Append to a file

```python
with open('file.txt', 'a') as f:
    f.write("New line\n")
```

11. Handle missing dictionary key

```python
try:
    d = {"name": "Alice"}
    print(d["age"])
except KeyError:
    print("Key not found")
```

12. Multiple except blocks

```python
try:
    x = int("abc")
except ValueError:
    print("Value error")
except TypeError:
    print("Type error")
```

13. Check if file exists

```python
import os
if os.path.exists("file.txt"):
    print("File exists")
else:
    print("No such file")
```

14. Log info and errors

```python
import logging
logging.basicConfig(filename='logfile.log', level=logging.INFO)
logging.info("This is an info message.")
logging.error("This is an error message.")
```

15. Print content and handle empty file

```python
with open('file.txt', 'r') as f:
    content = f.read()
    if not content:
        print("File is empty")
    else:
        print(content)
```

16. Memory profiling (requires memory_profiler)

```python
from memory_profiler import profile

@profile
def my_func():
    x = [i for i in range(10000)]
    return x

my_func()
```

17. Write list of numbers to file

```python
nums = [1, 2, 3, 4]
with open('numbers.txt', 'w') as f:
    for num in nums:
        f.write(f"{num}\n")
```

18. Logging with rotation (requires RotatingFileHandler)

```python
import logging
from logging.handlers import RotatingFileHandler

handler = RotatingFileHandler("rotating.log", maxBytes=1048576, backupCount=2)
logger = logging.getLogger()
logger.setLevel(logging.INFO)
logger.addHandler(handler)

logger.info("Log with rotation")
```

19. Handle IndexError and KeyError

```python
try:
    lst = [1, 2]
    print(lst[5])
    d = {"a": 1}
    print(d["b"])
except IndexError:
    print("Index out of range")
except KeyError:
    print("Key not found")
```

21. Count word occurrences

```python
word = "Python"
with open('file.txt', 'r') as f:
    content = f.read()
    count = content.count(word)
    print(f"{word} occurs {count} times.")
```

```python
import logging
logging.basicConfig(filename='file_error.log', level=logging.ERROR)
try:
    with open('nofile.txt', 'r') as f:
        content = f.read()
except FileNotFoundError as e:
    logging.error("File error: %s", e)
```
